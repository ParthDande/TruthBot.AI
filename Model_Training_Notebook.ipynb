```python
import re

def remove_numbers(text):
    # Use regex to replace digits with an empty string
    return re.sub(r'\d+', '', text)
def cleaningtitle(title):
    title=re.sub(r'\d+', '', title)
    title=title.lower()
    cleantitle = " ".join(cleanword for cleanword in title.split() if cleanword.isalnum())
    return cleantitle
```

```python
clean_df = df.copy()
clean_df["Title"]=df["Title"].apply(lambda x: cleaningtitle(x))
```

```python
import pandas as pd
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from sklearn.feature_extraction.text import TfidfVectorizer

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
```

```python
import nltk

# Check if 'punkt' and 'stopwords' are already downloaded
try:
    nltk.data.find('tokenizers/punkt')
    nltk.data.find('corpora/stopwords')
    print("NLTK data is already downloaded.")
except LookupError:
    print("NLTK data not found. Downloading...")
    nltk.download('punkt')
    nltk.download('stopwords')
```

```python
# Function to remove stopwords and tokenize
stop_words = set(stopwords.words('english'))
def preprocess_text(title):
    # Step 1: Tokenize and clean the title
    tokens = word_tokenize(title.lower())  # Tokenize the words in the title
    filtered_tokens = [word for word in tokens if word.isalnum() and word not in stop_words]  # Remove stopwords and non-alphanumeric tokens

    # Step 2: Join the filtered tokens back into a string
    combined_text = ' '.join(filtered_tokens)

    return combined_text  # Return the combined title + domain string

```

```python
chunk_size = 5000  # Process 5000 rows at a time
processed_chunks = []

for i in range(0, len(clean_df), chunk_size):
    chunk = clean_df.iloc[i:i + chunk_size]
    chunk['processed_text'] = chunk.apply(lambda row: preprocess_text(row['Title'] + ' ' + row['Text']), axis=1)
    processed_chunks.append(chunk)

# Combine all processed chunks
clean_df = pd.concat(processed_chunks, ignore_index=True)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Limit the features for TF-IDF to 20,000 most important ones
tfidf = TfidfVectorizer(max_features=20000, max_df=0.8, min_df=5)
title_vectors = tfidf.fit_transform(clean_df['processed_text'])
joblib.dump(tfidf, 'tfidf_vectorizer.pkl')

```

```python
import joblib

# Save preprocessed data
joblib.dump(clean_df, 'preprocessed_data.pkl')

# Save TF-IDF vectors
joblib.dump(title_vectors, 'tfidf_vectors.pkl')

# Reload when needed
clean_df = joblib.load('preprocessed_data.pkl')
title_vectors = joblib.load('tfidf_vectors.pkl')
```

```python
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    title_vectors,
    clean_df['label'],
    test_size=0.2,
    stratify=clean_df['label']
)

# Initialize SMOTE
smote = SMOTE(random_state=42)

# Fit and resample the training data
X_train_resampled, y_train_resampled = smote.fit_resample(X_train, y_train)
X_test_resampled,y_test_resampled = smote.fit_resample(X_test,y_test)
```

```python
def preprocess_custom_input(text,title):
    full_text = text+" + "+title
    # Step 1: Preprocess the text
    stop_words = set(stopwords.words('english'))
    tokens = word_tokenize(text.lower())  # Convert to lowercase and tokenize
    filtered_tokens = [word for word in tokens if word.isalnum() and word not in stop_words]  # Remove stopwords and non-alphanumeric tokens
    cleaned_text = ' '.join(filtered_tokens)
    # Step 2: Vectorize the cleaned text (using the same TF-IDF vectorizer)
    vectorized_text = tfidf.transform([cleaned_text])

    return vectorized_text
```

```python
def predict_fake_news(text, title,classifier):
    vector = preprocess_custom_input(text, title)
    # Get prediction and probabilities
    probabilities = classifier.predict_proba(vector)
    # Extract the predicted class (0 or 1) and confidence score
    predicted_class = probabilities[0].argmax()  # Get index of max probability
    confidence_score = probabilities[0][predicted_class]  # Confidence for the predicted class

    mapping = {1: "Real News", 0: "Fake News"}
    prediction = mapping[predicted_class]

    return prediction, confidence_score
```
